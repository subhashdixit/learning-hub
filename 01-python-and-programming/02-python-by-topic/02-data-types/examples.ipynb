# 2. Data Types

```python
# Integer
print(8)
# 1*10^308
print(1e309)
```

```python
# Decimal/Float
print(8.55)
print(1.7e309)
```

```python
# Boolean
print(True)
print(False)
```

```python
# Text/String
print('Hello World')
```

```python
# complex
print(5+6j)
```

```python
# List-> C-> Array
print([1,2,3,4,5])
```

```python
# Tuple
print((1,2,3,4,5))
```

```python
# Sets
print({1,2,3,4,5})
```

```python
# Dictionary
print({'name':'Nitish','gender':'Male','weight':70})
```
